```python
import re
from nltk.chat.util import Chat

# ============================================================
# CHATBOT-SYD 1.0
# BASE DE CONOCIMIENTO
# ============================================================


BASE = [

    # --------------------------------------------------------
    # REGLA 1: SALIR
    # --------------------------------------------------------
    (
        "salir",

        r"^\s*(?:salir|adi[oó]s|hasta luego)[.!?\s]*$",

        "Gracias por consultar Chatbot-SYD 1.0. "
        "Hasta luego."
    ),


    # --------------------------------------------------------
    # REGLA 2: CONSULTA VACÍA
    # --------------------------------------------------------
    (
        "vacio",

        r"^\s*$",

        "Escribe una consulta o escribe ayuda "
        "para ver los temas."
    ),


    # --------------------------------------------------------
    # REGLA 3: CONTACTAR A UN ASESOR
    # --------------------------------------------------------
    (
        "asesor",

        r".*\b(?:asesor|humano|contacto|seguimiento)\b.*",

        "Usa el formulario de contacto del sitio "
        "de la empresa. "
        "Indica producto o servicio, necesidad "
        "y presupuesto. "
        "Este prototipo no envía solicitudes "
        "ni registra datos."
    ),


    # --------------------------------------------------------
    # REGLA 4: GARANTÍA
    # --------------------------------------------------------
    (
        "garantia",

        r".*\b(?:garant[ií]a|garant[ií]as|devoluci[oó]n)\b.*",

        "Solicita al asesor la garantía aplicable, "
        "su duración, exclusiones y procedimiento. "
        "No hay una política cargada."
    ),


    # --------------------------------------------------------
    # REGLA 5: DISPONIBILIDAD
    # --------------------------------------------------------
    (
        "disponibilidad",

        r".*\b(?:disponibilidad|disponible|stock|existencias)\b.*",

        "Confirma con un asesor el producto, "
        "variante y cantidad. "
        "No tengo acceso al inventario en tiempo real."
    ),


    # --------------------------------------------------------
    # REGLA 6: PRECIO
    # --------------------------------------------------------
    (
        "precio",

        r".*\b(?:precio|precios|costo|cu[aá]nto cuesta)\b.*",

        "Consulta el precio vigente y solicita "
        "una cotización al asesor. "
        "Confirma impuestos y cargos adicionales; "
        "no tengo tarifas cargadas."
    ),


    # --------------------------------------------------------
    # REGLA 7: CONTRATACIÓN
    # --------------------------------------------------------
    (
        "contratacion",

        r".*\b(?:contratar|contrataci[oó]n|condiciones|comprar)\b.*",

        "Antes de comprar o contratar, confirma "
        "alcance, precio final, forma de pago "
        "y condiciones con la empresa. "
        "No proceso compras."
    ),


    # --------------------------------------------------------
    # REGLA 8: SELECCIÓN DE PRODUCTOS
    # --------------------------------------------------------
    (
        "seleccion",

        r".*\b(?:elegir|recomienda|recomiendas|presupuesto|opciones)\b.*",

        "Compara las opciones por uso previsto, "
        "características y presupuesto. "
        "Para recomendar un artículo concreto "
        "se necesita el catálogo y la validación "
        "de un asesor."
    ),


    # --------------------------------------------------------
    # REGLA 9: CARACTERÍSTICAS
    # --------------------------------------------------------
    (
        "caracteristicas",

        r".*\b(?:caracter[ií]sticas|detalles|cat[aá]logo)\b.*",

        "Revisa la ficha del producto o servicio: "
        "funciones, alcance y requisitos. "
        "Pide al asesor los detalles "
        "que no aparezcan allí."
    ),


    # --------------------------------------------------------
    # REGLA 10: HORARIO
    # --------------------------------------------------------
    (
        "horario",

        r".*\b(?:horario|horarios|atenci[oó]n)\b.*",

        "Puedo orientar mientras el programa "
        "esté abierto. "
        "Confirma el horario del equipo comercial "
        "en el sitio de la empresa."
    ),


    # --------------------------------------------------------
    # REGLA 11: AYUDA
    # --------------------------------------------------------
    (
        "ayuda",

        r".*\b(?:ayuda|men[uú])\b.*",

        "Temas: características, precio, "
        "disponibilidad, garantía, contratación, "
        "elegir, horario y asesor. "
        "Para terminar: salir."
    ),


    # --------------------------------------------------------
    # REGLA 12: SALUDO
    # --------------------------------------------------------
    (
        "saludo",

        r"^\s*(?:hola|buenos d[ií]as|buenas tardes)[.!?\s]*$",

        "Hola, soy Chatbot-SYD 1.0. "
        "Escribe ayuda para ver mis temas."
    ),


    # --------------------------------------------------------
    # REGLA 13: AGRADECIMIENTO
    # --------------------------------------------------------
    (
        "gracias",

        r"^\s*(?:gracias|muchas gracias)[.!?\s]*$",

        "Con gusto. "
        "Puedes hacer otra consulta o escribir salir."
    ),


    # --------------------------------------------------------
    # REGLA 14: CONSULTA DESCONOCIDA
    # --------------------------------------------------------
    (
        "desconocido",

        r"(?s)^.*$",

        "No tengo una regla para esa consulta. "
        "Escribe ayuda o asesor. "
        "Consulta un tema por mensaje."
    )
]


def crear_chatbot():
    #Convierte la base en pares de NLTK y crea el motor de reglas.
    pares = [(patron, [respuesta]) for _, patron, respuesta in BASE]
    return Chat(pares, reflections={})


def responder(chatbot, texto):
    #Entrega el texto original a NLTK, sin transformarlo.
    return chatbot.respond(texto)


def es_salida(texto):
    #Usa la misma regla de despedida para cerrar la conversación.
    return re.match(BASE[0][1], texto, flags=re.IGNORECASE) is not None


def main():
    #Mantiene consola hasta salir o cerrar la entrada.
    chatbot = crear_chatbot()
    print("Chatbot-SYD 1.0 | Orientación comercial General")
    print("Escribe ayuda para ver los temas o salir para terminar.")
    while True:
        try:
            texto = input("Cliente: ")
        except (EOFError, KeyboardInterrupt):
            print("\nSYD: Conversación finalizada.")
            break
        print("SYD:", responder(chatbot, texto))
        if es_salida(texto):
            break


if __name__ == "__main__":
    main()


```
